Conditional statements


Elif-else statements:



```python
attendance = 85
marks = 75

if attendance >= 75:
    if marks >= 35:
       print("Student is Eligible")
    else:
       print("Student Failed")
else:
    print("Attendance is Insufficient")
```

```python
salary = 30000

if salary >= 50000:
    bonus = 10000
elif salary >= 30000:
    bonus = 5000
elif salary >= 20000:
    bonus = 3000
else:
    bonus = 1000

print("Employee Bonus:", bonus)
```

Nested-if

```python
attendance = 85
marks = 75

if attendance >= 75:
    if marks >= 35:
        print("Student is Eligible")
    else:
        print("Student Failed")
else:
    print("Attendance is Insufficient")
```

```python
username = "admin"
password = "12345"

if username == "admin":
    if password == "12345":
        print("Login Successful")
    else:
        print("Incorrect Password")
else:
    print("Invalid Username")
```

Match-case statement

calculator

```python
a = 10
b = 5
operator = "+"

match operator:
    case "+":
        print("Addition:", a + b)
    case "-":
        print("Subtraction:", a - b)
    case "*":
        print("Multiplication:", a * b)
    case "/":
        print("Division:", a / b)
    case _:
        print("Invalid Operator")
```

Day of the week

```python
day = 3

match day:
    case 1:
        print("Monday")
    case 2:
        print("Tuesday")
    case 3:
        print("Wednesday")
    case 4:
        print("Thursday")
    case 5:
        print("Friday")
    case 6:
        print("Saturday")
    case 7:
        print("Sunday")
    case _:
        print("Invalid Day")
```

Looping statements

Types of looping statements: for loop and while programs

```python
names = ["Ram", "Ravi", "Krishna"]

for name in names:
    print(name)
 
 
for i in range(1, 6):
    print(i)
```

```python
i = 1

while i <= 5:
    print(i)
    i += 1
```

```python
i = 2

while i <= 4:
   print(i)
   i += 1
```

```python
##Nested Loop:

for i in range(1, 5):
   for j in range(1, 4):
       print(i, j)
 
```

Break and continue and programs

```python
for i in range(1, 6):
    if i == 4:
        break
    print(i)
```

```python
for i in range(1, 6):
    if i == 3:
        continue
    print(i)
```
