FUNCTIONS IN PYTHON

```python
def Bhavana():
    print("Hello I'm Bhavana")

Bhavana()
```

```python
marks = [85, 90, 78, 95, 88]

print("Total Marks:", sum(marks))
print("Highest Marks:", max(marks))
print("Lowest Marks:", min(marks))
print("Number of Subjects:", len(marks))
```

```python
def employee_details():
    name = "Bhavana"
    employee_id = 101
    salary = 25000

    print("Employee Name:", name)
    print("Employee ID:", employee_id)
    print("Employee Salary:", salary)

employee_details()
```

Arguments

```python
def addition(a, b):
    print(a + b)

addition(10, 20)
```

```python
def student(name, age, course):
    print("Student Name:", name)
    print("Age:", age)
    print("Course:", course)

student("Bhavana", 22, "Python")
```

```python
def employee(name, salary, department):
    print("Name:", name)
    print("Salary:", salary)
    print("Department:", department)

employee(department="IT", name="Bhavana", salary=30000)
```

```python
def shopping_bill(*prices):
    total = sum(prices)
    print("Total Shopping Bill:", total)

shopping_bill(500, 300, 200, 100)
```

```python
def welcome():
    print("Welcome to Python Training")

welcome()
```

```python
def get_salary():
    salary = 30000
    return salary

result = get_salary()

print("Employee Salary:", result)
```

```python
def addition(a, b):  
    return a + b

print(addition(10, 20))  
```

```python
def calculate_salary(basic, hra, bonus):
    total = basic + hra + bonus
    return total

salary = calculate_salary(20000, 5000, 3000)

print("Employee Total Salary:", salary)
```

```python
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 90:
        return "A Grade"
    elif average >= 75:
        return "B Grade"
    elif average >= 50:
        return "C Grade"
    else:
        return "Fail"

student_marks = [85, 90, 80, 95, 88]

print("Student Grade:", calculate_grade(student_marks))
```
